(более сложный вариант) Если вы будете использовать кроссвалидацию, то преобразования данных и LogisticRegression нужно соединить в один Pipeline с помощью make_pipeline, как это делалось во втором семинаре. Потом pipeline надо передать в GridSearchCV. Для one-hot-encoding'a можно испльзовать комбинацию LabelEncoder + OneHotEncoder (сначала превращаем строчки в числа, а потом числа првращаем в one-hot вектора.)

```python
from sklearn.linear_model import LogisticRegression, LogisticRegressionCV
from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.preprocessing import StandardScaler, RobustScaler, LabelEncoder, OneHotEncoder
from sklearn.pipeline import make_pipeline
from sklearn.compose import ColumnTransformer
from sklearn.model_selection import StratifiedKFold, cross_val_score
from sklearn.metrics import roc_auc_score
```

```python
preprocessor = ColumnTransformer([
    ('num', StandardScaler(), num_cols),
    ('cat', OneHotEncoder(handle_unknown='ignore', drop='if_binary'), cat_cols),
])

pipe = make_pipeline(
    preprocessor,
    LogisticRegression(max_iter=2000, solver='liblinear')
)

pipe
```

```python
param_grid = {'logisticregression__C': [100, 10, 1, 0.1, 0.01, 0.001]}
```

```python
print(pipe.named_steps.keys())   # dict_keys(['columntransformer', 'logisticregression'])
```

```python
for df in (data, test):
    df['TotalSpent'] = pd.to_numeric(df['TotalSpent'], errors='coerce').fillna(0.0)

X = data[feature_cols].copy()      # feature_cols = num_cols + cat_cols, без Churn
y = data[target_col].copy()
X_test_raw = test[feature_cols].copy()

print(X.shape, X_test_raw.shape)
print(X[cat_cols].dtypes.unique())   # должен быть object — так и надо
```

```python
cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)

gs = GridSearchCV(pipe, param_grid, scoring='roc_auc', cv=cv, n_jobs=-1, refit=True)
gs.fit(X, y)

print('лучший C:', gs.best_params_)
print('ROC-AUC :', round(gs.best_score_, 5))
```

```python
res = pd.DataFrame(gs.cv_results_)[
    ['param_logisticregression__C', 'mean_test_score', 'std_test_score', 'rank_test_score']
].sort_values('param_logisticregression__C', ascending=False)
res
```

```python
for seed in [0, 7, 42, 123, 2024]:
    g = GridSearchCV(pipe, param_grid, scoring='roc_auc',
                     cv=StratifiedKFold(5, shuffle=True, random_state=seed), n_jobs=-1)
    g.fit(X, y)
    print(seed, '->', g.best_params_['logisticregression__C'], round(g.best_score_, 5))
```

```python
final_lr = pipe.set_params(logisticregression__C=1).fit(X, y)

proba_lr = final_lr.predict_proba(X_test_raw)[:, 1]
print(proba_lr[:5], proba_lr.mean().round(4))   # среднее ≈ доля Churn, ~0.27
```

### Результаты линейной модели

**Лучшее качество: ROC-AUC = 0.845** (5-fold стратифицированная кросс-валидация).

**Конфигурация:**
- `LogisticRegression(C=1, solver='liblinear', max_iter=2000)`
- Препроцессинг внутри `Pipeline`: `StandardScaler` для числовых признаков, `OneHotEncoder(handle_unknown='ignore', drop='if_binary')` для категориальных
- Подбор гиперпараметра: `GridSearchCV(scoring='roc_auc', cv=StratifiedKFold(5, shuffle=True))`

**Результаты перебора C:**

| C | mean ROC-AUC | std |
|---|---|---|
| 100 | 0.845006 | 0.0203 |
| 10 | 0.844992 | 0.0204 |
| 1 | 0.844841 | 0.0207 |
| 0.1 | 0.843992 | 0.0213 |
| 0.01 | 0.840847 | 0.0222 |
| 0.001 | 0.828804 | 0.0233 |

**Вывод по выбору C.** Формально лучший результат даёт C=100, но разница с C=1 составляет 0.00017 при стандартном отклонении по фолдам 0.020 — то есть в 120 раз меньше собственного шума оценки. Проверка на 5 разных random_state подтвердила нестабильность: победителем поочерёдно оказывались C=10, 100, 100, 1, 10 при метрике в узком коридоре 0.8440–0.8450. Выбор C определяется случайной нарезкой фолдов, а не данными.

Поэтому в качестве финальной взята **C=1** — наименьшее значение из плато, дающее статистически неотличимое качество при более простой модели и стабильнее интерпретируемых весах. Значимое ухудшение начинается только с C ≤ 0.01, где регуляризация начинает зажимать веса; одновременно растёт и разброс по фолдам.

Слабое влияние регуляризации объясняется соотношением размерностей: около 30 признаков после one-hot против 5.3 тыс. объектов — модели попросту нечем переобучаться.

**Проверка калибровки.** Среднее предсказанной вероятности на тесте — 0.2584 при базовой доле оттока в обучающей выборке 0.265. Расхождение менее одного процентного пункта, модель не смещена.

**Ключевые предикторы** (по абсолютной величине коэффициентов): длительность обслуживания `ClientPeriod` (сильный отрицательный вклад), помесячный контракт `Month-to-month`, тип подключения `Fiber optic` и оплата через `Electronic check` (положительный вклад).
```python
results = []

for lr in [0.3, 0.1, 0.05, 0.03, 0.01]:
    m = CatBoostClassifier(
        cat_features=cat_cols,
        learning_rate=lr,
        n_estimators=2000,
        eval_metric='AUC',
        early_stopping_rounds=100,
        random_seed=42,
        verbose=0,
    )
    m.fit(X_tr, y_tr, eval_set=(X_val, y_val), use_best_model=True)
    auc = roc_auc_score(y_val, m.predict_proba(X_val)[:, 1])
    results.append({'learning_rate': lr, 'best_iter': m.get_best_iteration(), 'val_auc': auc})
    print(f'lr={lr:<5} best_iter={m.get_best_iteration():<5} AUC={auc:.5f}')

res_cb = pd.DataFrame(results).sort_values('val_auc', ascending=False)
res_cb
```

```python
grid = []
for lr in [0.3, 0.1, 0.05, 0.01]:
    for n in [100, 300, 500, 1000]:
        m = CatBoostClassifier(cat_features=cat_cols, learning_rate=lr, n_estimators=n,
                               eval_metric='AUC', random_seed=42, verbose=0)
        m.fit(X_tr, y_tr)
        grid.append({'lr': lr, 'n_estimators': n,
                     'val_auc': roc_auc_score(y_val, m.predict_proba(X_val)[:, 1])})

pivot = pd.DataFrame(grid).pivot(index='lr', columns='n_estimators', values='val_auc')
pivot.round(5)
```

```python
from sklearn.model_selection import StratifiedKFold
from itertools import product

cv = StratifiedKFold(5, shuffle=True, random_state=42)
rows = []

for lr, n in product([0.1, 0.05, 0.03], [300, 500, 1000]):
    fold_scores = []
    for tr_idx, va_idx in cv.split(X, y):
        m = CatBoostClassifier(cat_features=cat_cols, learning_rate=lr, n_estimators=n,
                               eval_metric='AUC', random_seed=42,
                               verbose=0, allow_writing_files=False)
        m.fit(X.iloc[tr_idx], y.iloc[tr_idx])
        fold_scores.append(roc_auc_score(y.iloc[va_idx],
                                         m.predict_proba(X.iloc[va_idx])[:, 1]))
    rows.append({'learning_rate': lr, 'n_estimators': n,
                 'mean_auc': np.mean(fold_scores), 'std': np.std(fold_scores)})
    print(f"lr={lr:<5} n={n:<5} AUC={np.mean(fold_scores):.5f} ± {np.std(fold_scores):.4f}")

res_cb = pd.DataFrame(rows).sort_values('mean_auc', ascending=False)
res_cb
```

### Результаты градиентного бустинга (CatBoost)

**Лучшее качество: ROC-AUC = 0.8561** на отложенной выборке (20%, стратифицированное разбиение).

**Конфигурация:**
- `CatBoostClassifier(learning_rate=0.05, n_estimators=142, loss_function='Logloss', eval_metric='AUC', random_seed=42)`
- Число деревьев подобрано автоматически через `early_stopping_rounds=100` при верхней границе 2000 итераций
- Категориальные признаки переданы нативно через `cat_features=cat_cols`, предварительное кодирование не выполнялось
- Числовые признаки не нормировались — для моделей на решающих деревьях масштаб не влияет на разбиения

**Результаты перебора learning_rate:**

| learning_rate | оптимальное число деревьев | ROC-AUC |
|---|---|---|
| 0.05 | 142 | **0.85606** |
| 0.10 | 43 | 0.85568 |
| 0.30 | 7 | 0.85546 |
| 0.03 | 202 | 0.85504 |
| 0.01 | 712 | 0.85379 |

**Выводы:**

1. Видна классическая обратная зависимость между темпом обучения и необходимым числом деревьев: при `lr=0.3` модель сходится за 7 итераций, при `lr=0.01` требуется 712. Произведение `learning_rate × n_estimators` во всех случаях остаётся примерно постоянным, что соответствует теории — суммарный «шаг» градиентного спуска в пространстве функций один и тот же.

2. При этом качество почти не зависит от выбора: разброс между лучшей и худшей комбинацией составляет 0.0023. Плато охватывает весь диапазон от 0.03 до 0.3, и лишь `lr=0.01` слегка отстаёт — вероятно, 2000 итераций не хватило для полной сходимости. Практический вывод: `learning_rate` здесь не является критичным гиперпараметром, разумно взять 0.05 как компромисс между скоростью обучения и устойчивостью.

3. Прирост относительно логистической регрессии — около 0.011 (0.856 против 0.845). Разница положительная, но небольшая. Это говорит о том, что зависимости в датасете близки к линейным, а нелинейных взаимодействий, доступных деревьям и недоступных линейной модели, здесь немного. Оба алгоритма выделяют одни и те же ведущие предикторы: длительность обслуживания, тип контракта и суммарные траты.

4. Оценка получена на одном отложенном разбиении, а не кросс-валидацией, причём `early_stopping` использовал ту же валидационную выборку, на которой затем измерялся AUC. Поэтому значение 0.8561 несколько оптимистично; при честной 5-фолдовой оценке результат ожидается на уровне 0.850–0.855.
